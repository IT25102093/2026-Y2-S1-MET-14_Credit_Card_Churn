```python
import warnings
warnings.filterwarnings('ignore')
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Load the dataset
df = pd.read_csv("BankChurners.csv")

# 2. Drop leakage and ID columns as required
leakage_cols = [c for c in df.columns if c.startswith("Naive_Bayes_Classifier")]
df = df.drop(columns=["CLIENTNUM"] + leakage_cols)

print("Dataset loaded successfully! Rows and columns:", df.shape)
```

```python
plt.figure(figsize=(12, 9))
numeric_df = df.select_dtypes(include=[np.number])
corr_matrix = numeric_df.corr()
sns.heatmap(corr_matrix, cmap='coolwarm', annot=False, square=True)
plt.title("Correlation Heatmap")
plt.show()
```
